```python
aapl_ticker = yf.Ticker('AAPL')
shares = aapl_ticker.get_shares_full(start='2021-10-01')
df_shares = pd.DataFrame(data=shares,  index=None, )
df_shares.reset_index(inplace=True)
df_shares = df_shares.rename(columns={'index': 'Date', 0: 'Shares'})
df_shares['Date'] = pd.to_datetime(df_shares['Date']).dt.date
df_shares.head()
print(type(df_shares['Date'][0]))
```

```python
price_history = aapl_ticker.history('5y')
price_history = price_history.reset_index()
price_history.head()
```

### Apple's P/B Ratio

```python
pb_ratios = []
market_caps = []
for year in balance:
    # Step 1: Find filing date
    file_date = year['filingDate']
    # Step 2: Find price on filing date
    filing_date_stock_price = price_history[price_history['Date'] == file_date]
    # Step 3: Convert to datetime
    file_date = pd.to_datetime(file_date)
    # Step 4: Find closest date available for share data
    df_shares['Date'] = pd.to_datetime(df_shares['Date'])
    closest_date = (df_shares['Date'] - file_date).abs().idxmin()
    closest_row = df_shares.loc[closest_date]
    # Step 5: Find market cap
    market_cap = filing_date_stock_price['Close'].iloc[0] * closest_row['Shares']
    market_cap = float(market_cap)
    market_caps.append(market_cap)
    # Step 6: Find P/B ratio
    shareholder_equity = year['totalStockholdersEquity']
    ratio = market_cap / shareholder_equity
    pb_ratio = float(ratio)
    pb_ratios.append(pb_ratio)

print(pb_ratios)
print(market_caps)
```

### Apple's D/E Ratio

```python
de_ratios = []

for year in balance:
    # Step 1: Pull shareholder and total debt from balance sheet
    shareholder_equity = year['totalStockholdersEquity']
    total_debt = year['totalDebt']
    # Step 2: Divide numbers
    de_ratio = total_debt / shareholder_equity
    # Step 3: Append yearly ratios to list
    de_ratios.append(de_ratio)

print(de_ratios)
```

### Altman Z-score for manufacturing firms

### Altman Z-score for nonmanufacturing

```python
# Altman Z-Score = 6.56A + 3.26B + 6.72C + 1.05D

# Where:

# A = working capital (current assets − current liabilities) / total assets
# B = retained earnings / total assets
# C = earnings before interest and taxes / total assets
# D = book value of equity / total liabilities

altman_z_score = []

for x in range(0, 5):
    # Find total assets and total liabilities
    total_assets = balance[x]['totalAssets']
    total_liabilities = balance[x]['totalLiabilities']
    # Find current assets and current liabilities
    current_assets = balance[x]['totalCurrentAssets']
    current_liabilities = balance[x]['totalCurrentLiabilities']
    # Calculate working capital and A
    working_capital = current_assets - current_liabilities
    A = working_capital / total_assets
    A *= 6.56 
    print(A)
    # Calculate B
    retained_earnings = balance[x]['retainedEarnings']
    B = (retained_earnings / total_assets)
    B *= 3.26
    print(B)
    # Calculate C
    ebit = income[x]['ebit']
    C = ebit / total_assets
    C *= 6.72
    print(C)
    # Calculate D
    D = balance[x]['totalStockholdersEquity'] / total_liabilities
    D *= 1.05
    print(D)

    score = A + B + C + D
    altman_z_score.append(score)

print(altman_z_score)
```
